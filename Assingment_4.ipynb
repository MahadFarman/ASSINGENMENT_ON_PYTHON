#### Q4 — Count Frequency of Numbers
<pre>
numbers = [1,2,2,3,3,3,4,4,4,4]
<pre>
Create frequency dictionary.

```python
numbers =[1,2,2,3,3,3,4,4,4,4]
frequency = {}

for num in numbers:
    frequency[num] = numbers.count(num)
    
print(frequency)
```

#### Q5 — Word Counter
<pre>
Take sentence input and count word frequency.

```python
sentence = input("Enter the sentence :")
words = sentence.split()

word_count = {}

for word in words:
    word_count[word] =words.count(word)

print(f"Your sentence is :{sentence}")
print(word_count)
```

#### Q6 — Student Marks & Average
<pre>
marks = {"Math":80, "English":70, "Physics":90}
<pre>
Calculate total and average.

```python
marks = {
    "Math":80,
    "English":70,
    "Physics":90
    }

total_marks = sum(marks.values())
avg = total_marks / 3

print(f"Your total marks is:{total_marks}")
print(f"Your average marks is:{avg}")
```

#### Q7 — Highest Scoring Student
<pre>
students = {
"Ali":85,
"Sara":92,
"Ahmed":78,
"Zara":95
}
<pre>
Find highest scoring student.

```python
students ={
    "Ali":85,
    "Sara":92,
    "Ahmed":78,
    "Zara":95
    }

highest_marks = max(students, key=students.get)

print(f"Highest scoring student:{highest_marks}" )
print(f"Marks:{students[highest_marks]}")
```

#### Q8 — Merge Dictionaries
<pre>
dict1 = {"a":1,"b":2}
dict2 = {"c":3,"d":4}
<pre>
Merge and print result.

```python
dict_1 = {
    "a":1,
    "b":2
    }

dict_2 = {
    "c":3,
    "d":4
    }

merged_dict = dict_1.copy()
merged_dict.update(dict_2)

print(merged_dict)
```
