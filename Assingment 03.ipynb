#### 1. Write a program that accepts a list from user and print the alternate element of list.

```python
l = []

for i in range(6):
    n = int(input("Enter the numbers :"))
    l.append(n)

print(f"Your numbers are {l}")
print("Your alternative numbers are", l[0::2])
```

#### 2. Write a program that accepts a list from user. Your program should reverse the content of list and display it. Do not use reverse() method.  

```python
l = []

for i in range(6):
    n = int(input("Enter the numbers :"))
    l.append(n)

print(f"Your numbers are :{l}") 
print("Your reverse numbers are :",l[5::-1])   
```

#### 3. Find and display the largest number of a list without using built-in function max(). Your program should ask the user to input values in list from keyboard.  

```python
l = []

for i in range(6):
    n = int(input("Enter the numbers :"))
    l.append(n)

print(f"Your numbers are:{l}")

largest =l[0]

for i in l:
    if i > largest:
        largest = i

print(f"Your largest number is: {largest}") 
```

#### 6. Write a program that reads a string from the user containing a date in the form mm/dd/yyyy. It should print the date in the form March 12, 2021

```python
date = input("Enter date (mm/dd/yyyy) :")
month, day, year = date.split("/")

months ={
    "01": "January",
    "02": "February",
    "03": "March",
    "04": "April",
    "05": "May",
    "06": "June",
    "07": "July",
    "08": "August",
    "09": "September",
    "10": "October",
    "11": "November",
    "12": "December" 
}

print(months[month], day +",", year)

```

#### 9. Write a program to add two matrices of size n x m. 

```python
matrix_1 = [
    [2, 11],
    [5, 2]
]

matrix_2 = [
    [1, 3],
    [4, 6]
]
result = [
    [matrix_1[0][0] + matrix_2[0][0], matrix_1[0][1] + matrix_2[0][1]],
    [matrix_1[1][0] + matrix_2[1][0], matrix_1[1][1] + matrix_2[1][1]]
]
print(result)
```

#### 10. Write a program to multiply two matrices

```python
matrix_1 = [
    [2, 3],
    [4, 5]
]

matrix_2 = [
    [1, 2],
    [3, 4]
]

result = [
    [matrix_1[0][0] * matrix_2[0][0] + matrix_1[0][1] * matrix_2[1][0],
     matrix_1[0][0] * matrix_2[0][1] + matrix_1[0][1] * matrix_2[1][1]],

    [matrix_1[1][0] * matrix_2[0][0] + matrix_1[1][1] * matrix_2[1][0],
     matrix_1[1][0] * matrix_2[0][1] + matrix_1[1][1] * matrix_2[1][1]]
]

print(result)
```
