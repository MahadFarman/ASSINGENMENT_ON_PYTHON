9) Using groupby, find the average salary per City, the number of employees per City, and the maximum salary per City. Show the result using agg() with multiple functions.

```python
data = {
    "Employee_ID": [101, 102, 103, 104, 105, 106, 107, 108],
    "First_Name": ["Mahad", "Affan", "Sarim", "Ahad","Mustafa", "Ali", "Rehan", "Azkia"],
    "Last_Name": ["Farman", "Khan", "Ahmed", "Ali","Raza", "Malik", "Ahmed", "Rauf"],
    "Department": ["Sales", "IT", "HR", "IT","Sales", "HR", "IT", "Sales"],
    "Salary": [85000, 65000, 62000, 78000,70000, 65000, 95000, 80000],
    "City": ["Karachi", "Lahore", "Karachi", "Islamabad","Lahore", "Karachi", "Islamabad", "Lahore"]
}

df = pd.DataFrame(data)
df
```

```python
# Group by City and apply multiple functions

Analysis = df.groupby("City").agg(
    Avg_Salary=("Salary", "mean"),
    No_of_Employees=("Employee_ID", "count"),
    Max_Salary=("Salary", "max")
)
Analysis
```

10) Create two DataFrames (employees and departments) with a common column 'DeptID'. Merge them using inner and left joins, then concatenate two DataFrames of the same structure using pd.concat() and remove duplicate rows.

```python
employees = pd.DataFrame({
    "id":         [1, 2, 3, 4],
    "name":       ["Mahad", "Affan", "Ahad", "Ali"],
    "dept_id":    [10, 20, 10, 30]
})

departments = pd.DataFrame({
    "dept_id":   [10, 20, 40],
    "dept_name": ["Engineering", "Finance", "Marketing"]
})
print(employees)
print(departments)
```

```python
inner = pd.merge(employees, departments, on="dept_id", how="inner")
inner
```

```python
left = pd.merge(employees, departments, on="dept_id", how="left")
left
```

```python
combined = pd.concat([employees, employees], ignore_index=True)
combined = combined.drop_duplicates()

print(combined)
```
