14)A student will not be allowed to sit in exam if his/her attendence is less than 75%.

Take following input from user

- Number of classes held

- Number of classes attended.

And print

- percentage of class attended

- Is student is allowed to sit in exam or not.


```python
classes_held = int(input("Enter number of classes_held :"))

classes_attended= int(input("Enter number of class_attended :"))

print(f"Number of classes_held: {classes_held}")
print(f"Number of classes_attended: {classes_attended}")

total_attendence = classes_attended / classes_held * 100

print(f"Your total_attendence is: {total_attendence}")

if total_attendence >= 75 :
    print("Your are allow to sit in exam")

else:
    print("Your are not allow to sit in exam")

```

16) Write a program to check if a year is leap year or not.

If a year is divisible by 4 then it is leap year but if the year is century year like 2000, 1900, 2100 then it must be divisible by 400.

```python
Year = int(input("Enter the year"))

print(f"The year is {Year}")

if Year % 400 ==0 :
    print("It's a leap year")

elif Year % 100 == 0:
    print("It's not a leap year")

elif Year % 4 == 0:
    print("It's a leap year")

else:
    print("It's not a leap year")

```
