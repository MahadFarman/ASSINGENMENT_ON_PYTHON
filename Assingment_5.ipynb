## Q1: Working With Lists

```python
numbers = [12, 45, 87, 90, 23 ,22, 67, 88]

print(f"All numbers of this list is {numbers}")

sum = 0

for n in numbers:
    sum = sum + n
print(f"The sum of these numbers is {sum}")

for n in numbers:
    if n % 2 == 0:
        print(f"The even numbers in this lists is {n}")
```

## Q2: Dictionary Basics

```python
dictionary = {
    "Name" : "Mahad",
    "Age" : 16,
    "Course" : "Data Analytics With Python",
    "Marks" : [94, 60, 55]
}

for dict in dictionary:
    print(dict, dictionary[dict])

sum = 0

for i in dictionary["Marks"]:
    sum = sum + i
    avg = sum / 3
print(f"The average of these marks is :{avg}")

if avg >= 50:
    print("You passed")

else:
    print("You failed")
```

## Q3: Lists of dictionaries

```python
employees = []

employee_1 = {
    "name": "Mahad",
    "department": "IT",
    "salary": 110000
}

employee_2 = {
    "name": "Affan",
    "department": "HR",
    "salary": 80000
}

employee_3 = {
    "name": "Ahad",
    "department": "Finance",
    "salary": 75000
}

employees = [employee_1, employee_2, employee_3]

highest_salary = 0

for i in employees: 
    print(i)

    if i["salary"] > highest_salary:
        highest_salary = i["salary"]
print(f"The highest_salary of employee is: {highest_salary}" )

total_salary = 0

for i in employees:
    total_salary = total_salary + i["salary"]
print(f"The total_salary of these employees is: {total_salary}")

```

## Q4: While loop with user input

```python
num =[]

while True:
    numbers = int(input("Enter the number :"))
    if numbers == -1: 
        break
    num.append(numbers)
print(f"Your complete list is :{num}")

largest_num = num[0]
smallest_num = num[0]

for n in num:
    if n > largest_num:
        largest_num = n

    if n < smallest_num:
        smallest_num = n

print(f"Your largest number is: {largest_num}")
print(f"Your smallest number is: {smallest_num}")
```

## Q5: Word frequency counter

```python
sentence = input("Enter the sentence :")

words = sentence.split()

words_count = {}

print(sentence)

for word in words:
    if word in words_count:
        words_count[word]  += 1

    else:
        words_count[word]  = 1

for word in words_count:
    count = words_count[word]
    print(f"Your word is :{word} \n  Its count is :{count}")


```

## Q6: Nested loops - Multiplication tables

```python
# Multiplication table from 1 to 5

# First list
numbers = [1, 2, 3, 4, 5]

# Second list
table_rows = []

# First dictionary
multiplication_table = {}

# Second dictionary
saved_table = {}

# First loop
for number in numbers:
    row = []
    # Nested loop
    for i in numbers:
        result = number * i
        row.append(result)

    table_rows.append(row)

    multiplication_table[number] = row


# Second loop
for row in table_rows:
    # Nested loop
    for value in row:
        print(value, end=" ")

    print()


# Third loop
for number in multiplication_table:
    saved_table[number] = multiplication_table[number]
```
